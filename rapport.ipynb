Machine Learning: vin-prediksjon

Problemet
Målet med dette prosjektet er å utvikle en maskinlæringsmodell som kan predikere kvaliteten på vin basert på kjemiske verdier. Applikasjonen gjør det mulig for brukeren å legge inn forskjellige kjemiske komponenter fra vinen og motta en predikert kvalitetsscore.

Kvaliteten på vin blir vanligvis evaluert gjennom sensoriske vurderinger av mennesker, enten eksperter eller forbrukere [1]. Vurderingen er derfor individuell og kan variere fra person til person. Kjemiske analyser gir oss målbar informasjon om vinen, og maskinlæring kan analysere disse verdiene og sammenligne dem med menneskelige vurderinger.

Prosjektet er gjennomført med supervised machine learning, hvor en modell er trent til å tolke mønstre fra tidligere evaluerte viner. Gjennom læring fra kjemiske verdier og kvalitetsscorer gitt av mennesker vil modellen kunne gi et estimat på hvor god vinen er.

Den tiltenkte brukeren er alle som lager, kjøper, selger, investerer i eller drikker vin. Applikasjonen er hovedsakelig laget for å demonstrere hvordan maskinlæring kan brukes til å predikere kvaliteten på vin.

Uten maskinlæring ville vinkvalitet normalt blitt vurdert gjennom smak, lukt, utseende, laboratorieundersøkelser og vurderinger fra eksperter som ser på helheten. Maskinlæringsmodellen vil kunne se på mønstre fra tidligere vurderinger og gi en vurdering basert på dette.

Den endelige løsningen består av to hovedkomponenter:
En maskinlæringsmodell trent på vindata hentet fra UCI Machine Learning [2].
En webapplikasjon utviklet ved bruk av Streamlit, hvor brukeren kan legge inn kjemiske verdier for vinen og få en prediksjon.

Applikasjonen er i dag laget for å kunne kjøres lokalt på en PC.

Metrikker
Hvor godt maskinlæringsmodellene presterer, vurderes basert på tre metrikker: Mean Absolute Error (MAE), Root Mean Squared Error (RMSE) og R-kvadrert (R²).
Mean Absolute Error måler gjennomsnittlig absolutt differanse mellom den predikerte kvaliteten og den faktiske kvaliteten. En lav MAE-verdi indikerer at prediksjonen ligger nær den faktiske verdien.
Root Mean Squared Error måler prediksjonsfeil, samtidig som den vektlegger større feil mer enn mindre feil. Den er nyttig for å fange opp større feil som maskinlæringsmodellen noen ganger kan gjøre.
R-kvadrert måler hvor stor del av variasjonen i vinkvalitet som forklares av modellen, sammenlignet med en modell som alltid forutsier gjennomsnittlig kvalitetsscore.
MAE brukes som utgangspunkt for å avgjøre hvilken modell som presterer best, fordi den er lettest å tolke. Resultatene representerer den gjennomsnittlige prediksjonsfeilen i kvalitetspoeng. RMSE og R² gir i tillegg informasjon om ytelsen til modellene.
Prosjektet anses som vellykket fordi det har en applikasjon som tar imot verdier brukeren taster inn, og gir en prediksjon basert på disse inputverdiene.

Data
Prosjektet benytter datasett for vin hentet fra UCI Machine Learning [2].
Datasettet inneholder kjemiske målinger av vinprøver, samt kvalitetsscorer gitt av menneskelige vinsmakere. Det består av to datasett: ett for rødvin og ett for hvitvin.
Hver observasjon inneholder 11 inndatavariabler (features):
Fast syre (Fixed acidity)
Flyktig syre (Volatile acidity)
Sitronsyre (Citric acid)
Rest-sukker (Residual sugar)
Klorider (Chlorides)
Fri svoveldioksid (Free sulfur dioxide)
Total svoveldioksid (Total sulfur dioxide)
Tetthet (Density)
pH
Sulfater (Sulphates)
Alkohol (Alcohol)
Disse dataene er de samme som applikasjonen ber om fra brukeren.

Målvariabelen er «kvalitet», som representerer kvalitetsscoren som er tildelt hver vin. Kvalitetsscoren er et heltall basert på sensoriske vurderinger. Disse scorene fungerer som fasit under supervised learning.
Datasettet er offentlig tilgjengelig og inneholder kjemiske målinger og kvalitetsvurderinger av vin, ikke personopplysninger. Prosjektet behandler derfor ingen personopplysninger.

Klargjøring av data
Datasettene klargjøres og hentes inn ved hjelp av biblioteket Pandas. Inndatavariablene (X) blir skilt fra målvariabelen (y).
Datasettene deles inn i et treningssett og et testsett, med en fordeling på 80/20. Det vil si at 80 % av observasjonene brukes til trening av modellene, mens de resterende 20 % brukes til testing og evaluering. En fast tilfeldig tilstand (random state) benyttes for å sikre reproduserbarhet og sørge for at alle modellene evalueres på de samme observasjonene.
Treningssettet brukes til å trene modellene, mens testsettet brukes senere til å evaluere ytelsen på observasjoner modellen ikke har sett tidligere.
For Linear Regression benyttes skalering av variabler (feature scaling). Dette sikrer at variabler med ulike numeriske skalaer blir standardisert. For modeller basert på beslutningstrær, som Random Forest og Gradient Boosting, er slik skalering normalt unødvendig.
Det opprinnelige datasettet brukes uten å fjerne noen av de 11 kjemiske inndatavariablene eller legge til ytterligere manuelt utformede variabler.

Modellering
Modellvalg
Tre modeller for supervised learning ble undersøkt:
1. Lineær regresjon (Linear Regression)
Lineær regresjon brukes som en grunnmodell (baseline-modell). Den forsøker å finne en lineær sammenheng mellom inndatavariablene og målvariabelen. Modellen estimerer koeffisienter som beskriver hvordan de ulike kjemiske egenskapene bidrar til den predikerte vinkvaliteten. Den er relativt enkel og lett å tolke, men kan ha begrensninger dersom sammenhengene mellom variablene og vinkvaliteten er mer komplekse og ikke-lineære.

2. Random Forest
Random Forest er en ensemble-læringsmetode basert på beslutningstrær. I stedet for å stole på ett enkelt beslutningstre oppretter modellen mange trær og kombinerer prediksjonene deres. Hvert tre lærer ulike mønstre fra treningsdataene. For regresjonsoppgaver beregnes den endelige prediksjonen ved å ta gjennomsnittet av prediksjonene fra de enkelte trærne. Random Forest kan fange opp ikke-lineære sammenhenger og samspill mellom inndatavariabler. Den er også relativt robust mot overtilpasning (overfitting), sammenlignet med et enkelt beslutningstre.
Denne implementeringen benytter 200 beslutningstrær.

3. Gradient Boosting
Gradient Boosting er en annen ensemble-læringsmetode basert på beslutningstrær. I motsetning til Random Forest, der trærne bygges uavhengig av hverandre, bygges trærne i Gradient Boosting sekvensielt. Hvert nye tre forsøker å korrigere feil som ble gjort av de foregående trærne. Dette gjør at modellen gradvis kan forbedre prediksjonene sine ved å lære av tidligere prediksjonsfeil. Gradient Boosting kan fange opp komplekse sammenhenger, men kan kreve mer nøyaktig justering av parametere.

Trening og evaluering av modeller
Alle de tre modellene trenes og evalueres ved hjelp av de samme trenings- og testdatasettene for å sikre en rettferdig sammenligning.
I tillegg til en 80/20-oppdeling mellom trening og test benyttes femdelt kryssvalidering (five-fold cross-validation) på treningsdataene.
Ved femdelt kryssvalidering deles treningsdataene inn i fem delsett. Modellen trenes og valideres flere ganger, der ulike delsett brukes til validering. Gjennomsnittlig valideringsfeil gir et mer pålitelig estimat av modellens ytelse.
MAE (gjennomsnittlig absolutt feil) fra kryssvalideringen brukes som hovedkriterium for valg av den endelige modellen. Testsettet brukes deretter til å evaluere den valgte modellens ytelse på observasjoner den ikke har sett tidligere.

Variabelviktighet (Feature Importance)
Random Forest gir et mål på variabelviktighet, som indikerer hvor nyttige de enkelte inndatavariablene var ved konstruksjonen av beslutningstrærne.
Dette kan bidra til å identifisere hvilke kjemiske egenskaper som bidro mest til modellens prediksjoner. Variabelviktighet kan gi innsikt i modellens virkemåte, selv om det ikke beviser at en bestemt kjemisk egenskap direkte forårsaker høyere eller lavere vinkvalitet.
Begrensninger
Modellen er trent på historiske vurderinger av vinkvalitet. Siden disse scorene stammer fra menneskelige sensoriske evalueringer, kan de inneholde subjektiv variasjon.
Modellen predikerer derfor kvalitetsscorer basert på mønstre i det eksisterende datasettet, snarere enn å måle objektiv vinkvalitet. Videre er det ikke sikkert at modellen fungerer like godt på viner som skiller seg vesentlig fra prøvene som inngår i treningsdataene.
Implementering
Den endelige maskinlæringsmodellen er implementert gjennom en lokal webapplikasjon utviklet med Streamlit.
Etter at modellen er trent, lagres den ved hjelp av biblioteket Joblib. Dette gjør at applikasjonen kan laste inn den trente modellen uten å måtte trene den på nytt hver gang nettsiden startes.
Applikasjonen har inntastingsfelt som tilsvarer de 11 kjemiske egenskapene som ble brukt under treningen. Når brukeren sender inn verdiene, organiserer applikasjonen dem i riktig format og sender dem videre til den trente Random Forest-modellen.
Modellen returnerer en estimert poengsum for vinkvalitet, som vises for brukeren. Applikasjonen fungerer dermed som et enkelt grensesnitt mellom maskinlæringsmodellen og sluttbrukeren.
Vedlikehold av modellen og fremtidige forbedringer
Modellen kan trenes på nytt når flere vindata blir tilgjengelige, eller når det identifiseres forbedringsmuligheter for modellen.
Mulige fremtidige forbedringer inkluderer:
•	Testing av ytterligere maskinlæringsalgoritmer.
•	Gjennomføring av hyperparameteroptimalisering.
•	Undersøkelse av variabelviktighet og prediksjonsfeil.
•	Forbedring av brukergrensesnittet og tilføyelse av visualiseringer.
